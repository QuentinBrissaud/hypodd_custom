```python
import pandas as pd
import numpy as np
```

```python
file = './data/starting-station-Malyn.csv'
station_df = pd.read_csv(file, header=[0],)

file = './data/starting-event-Malyn.csv'
event_df = pd.read_csv(file, header=[0],)

file = './data/starting-phase-Malyn.csv'
phase_df = pd.read_csv(file, header=[0],)
```

```python
from csv_to_obspy_xml import convert_dataframes_to_xml

catalog, inventory = convert_dataframes_to_xml(
    event_df=event_df,
    station_df=station_df,
    phase_df=phase_df,
    event_xml="events.xml",
    station_xml="stations.xml",
    fixed_depth_km=0.01,  # optional
)
```

```python
from obspy import read_events, read_inventory

catalog = read_events("events.xml")
inventory = read_inventory("stations.xml")
```
