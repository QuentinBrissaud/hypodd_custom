# Relocation Vector Explorer

This notebook loads the same HypoDD outputs used by `hypodd_quality_report.py` and draws an interactive Matplotlib version of `ukraine_relocation_vectors_map.png`. Use the Matplotlib toolbar to zoom/pan, and use the dropdown to recolor relocated events.

```python
# For zoom/pan inside JupyterLab, this prefers ipympl and falls back to notebook mode.
try:
    get_ipython().run_line_magic("matplotlib", "widget")
except Exception:
    get_ipython().run_line_magic("matplotlib", "notebook")

from pathlib import Path

import matplotlib.pyplot as plt

from relocation_vector_explorer import (
    available_color_columns,
    create_relocation_vector_interface,
    load_relocation_vector_dataframe,
    load_station_dataframe,
    plot_relocation_vectors,
)
```

Set this to either the relocation working directory or directly to its `output_files` directory.

```python
working_dir = Path("relocation_test_iterations2_damp5_nocc_clusters")

df = load_relocation_vector_dataframe(working_dir)
stations = load_station_dataframe(working_dir)

print(f"Loaded {len(df)} relocated events")
print("Available color columns:")
for column in available_color_columns(df):
    print(" -", column)

df.head()
```

Interactive version. Useful color choices are usually `cluster_id`, `rms_ct_s`, and `rms_cc_s` when cross-correlation data are present.

```python
viewer = create_relocation_vector_interface(
    working_dir,
    default_color_by="cluster_id",
    default_basemap="natural_earth",  # use "osm" for OpenStreetMap tiles
    osm_zoom=9,
    use_cartopy=True,
)
```

Static/reproducible call. Re-run this cell with a different `color_by` value if you do not want to use widgets.

```python
fig, ax, scatter = plot_relocation_vectors(
    df,
    stations=stations,
    color_by="rms_ct_s",
    basemap="osm",
    osm_zoom=9,
    use_cartopy=True,
)
plt.show()
```
